# S4: Hybrid pipeline (Most Efficient)

**Stage A** — Multi-rotation NCC template matching at a *loose* threshold
(0.50). Deliberately over-generates candidates to protect recall.

**Stage B** — DINOv3 cosine-similarity rerank of each candidate (crops
upsampled to 224×224, batched). Final score = weighted combo of NCC and
cosine; accept above a recall-favored threshold.

**Stage C** — On pages where Stage A returned almost nothing, fire a
DINOv3 sliding-window scan as the false-negative safety net. Conditional,
not run on every page — keeps wall-clock tractable.

```python
import sys, os
from pathlib import Path
PKG_ROOT = Path.cwd().parents[0] if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PKG_ROOT / 'src'))
os.chdir(PKG_ROOT)

import warnings
warnings.filterwarnings('ignore')

import numpy as np, cv2, matplotlib.pyplot as plt
from IPython.display import display
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['figure.dpi'] = 80
```

```python
from symbol_matching import benchmark, pdf_io, viz, embed
from symbol_matching.strategies import s4_hybrid
from symbol_matching.eval import evaluate
from symbol_matching.records import dump_matches
print('embedding model:', embed.model_name())

ref = benchmark.load_reference_crop()
pages = [pdf_io.render_page(benchmark.CD_SET_PDF, i, dpi=benchmark.DPI) for i in range(3)]
metas = [pdf_io.extract_metadata(benchmark.CD_SET_PDF, i) for i in range(3)]
gts = benchmark.ground_truth_per_page()
```

## Run S4 on all three pages

```python
import time, os
os.makedirs('outputs/s4', exist_ok=True)
all_matches = []
for pi in range(3):
    meta = metas[pi]
    t0 = time.time()
    matches = s4_hybrid.find_matches_on_page(
        pages[pi], ref,
        stage_a_threshold=0.50,
        stage_b_accept_threshold=0.65,
        stage_c_min_proposals=10,
        stage_c_threshold=0.85,
        page_idx=pi,
        page_meta={'sheet_ref': meta.sheet_ref, 'page_name': meta.page_name, 'page_type': meta.page_type},
    )
    dt = time.time() - t0
    e = evaluate(matches, gts[pi])
    fb = sum(1 for m in matches if m.item.stage == 'fallback')
    print(f'p{pi} ({meta.sheet_ref}): {len(matches)} matches ({fb} from Stage C) / GT={len(gts[pi])}  '
          f'P={e.precision:.2f}  R={e.recall:.2f}  F1={e.f1:.2f}  {dt:.1f}s')
    all_matches.append(matches)
    viz.save_annotated(pages[pi], matches, f'outputs/s4/p{pi}_annotated.png', thickness=4, font_scale=0.6)
```

## Output records (drawingItem + capture JSON)

```python
flat = [m for ms in all_matches for m in ms]
dump_matches(flat, 'outputs/s4/matches.json')
print(f'wrote {len(flat)} matches to outputs/s4/matches.json')
# Show one record
import json
print(json.dumps(json.loads(open('outputs/s4/matches.json').read())[0], indent=2))
```
