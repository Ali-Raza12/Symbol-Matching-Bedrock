# S3: Pure DINOv3 sliding-window matching

A pure-embedding alternative to template matching. For each page:

1. Slide a window of the reference's pixel size across the page at a
   small stride.
2. Batch-encode all windows with DINOv3 (after upsampling each window
   to the encoder's 224×224 input).
3. Cosine-similarity each window's embedding against the reference's
   embedding (max over 4 cardinal rotations).
4. Threshold the heatmap, NMS, emit matches.

This catches stylistic variation between instances that breaks pixel
correlation. The cost is wall-clock: many windows × DINOv3 forward pass.

```python
from symbol_matching import benchmark, pdf_io, embed, viz
from symbol_matching.strategies import s3_dinov3
from symbol_matching.eval import evaluate
print('embedding model:', embed.model_name())
ref = benchmark.load_reference_crop()
pages = [pdf_io.render_page(benchmark.CD_SET_PDF, i, dpi=benchmark.DPI) for i in range(3)]
gts = benchmark.ground_truth_per_page()
```

## Sanity: cosine similarity sanity check

Before running the matcher, confirm that DINOv3 separates real callouts
from non-callouts when crops are upsampled to 224×224 (the encoder's
preferred input size).

```python
ref_emb = embed.encode_crop(ref)
sample_gt = gts[0][:8]

crops = []
for b in sample_gt:
    crop = pages[0][max(0,b.y-6):b.y+b.h+6, max(0,b.x-6):b.x+b.w+6]
    crops.append(crop)
crops.append(pages[0][:200, :200]) #blank region          
crops.append(pages[0][800:1100, 8500:8800])  # text region
labels = ['GT'] * len(sample_gt) + ['blank', 'text']

embs = embed.encode_crops(crops)
sims = embs @ ref_emb
for label, s in zip(labels, sims):
    print(f'  {label:5s}  cosine={s:.3f}')
```

## Sliding-window matcher on page 0

```python
import time
t0 = time.time()
matches_p0 = s3_dinov3.find_matches_on_page(pages[0], ref, stride_frac=0.5, threshold=0.80, page_idx=0)
dt = time.time() - t0
e = evaluate(matches_p0, gts[0])
print(f'p0: {len(matches_p0)} matches / GT={len(gts[0])}  P={e.precision:.2f}  R={e.recall:.2f}  F1={e.f1:.2f}  {dt:.1f}s')
```
